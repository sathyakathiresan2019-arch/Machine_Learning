```python
class QuotationCalculator:

    def __init__(self):

        # Material Rates (₹)
        self.cement_rate = 450
        self.sand_rate = 80
        self.aggregate_rate = 120
        self.steel_rate = 75
        self.brick_rate = 8

        # Charges
        self.transport_percent = 2
        self.safety_percent = 1
        self.gst_percent = 18
        self.profit_percent = 10

    def calculate(
        self,
        description,
        quantity,
        cement,
        sand,
        aggregate,
        steel,
        bricks,
        water,
        excavator,
        hire,
        labour,
        helper,
        electrical
    ):

        # Material Cost
        cement_cost = cement * self.cement_rate
        sand_cost = sand * self.sand_rate
        aggregate_cost = aggregate * self.aggregate_rate
        steel_cost = steel * self.steel_rate
        brick_cost = bricks * self.brick_rate

        material_cost = (
            cement_cost +
            sand_cost +
            aggregate_cost +
            steel_cost +
            brick_cost +
            water
        )

        # Labour
        labour_cost = labour + helper

        # Equipment
        equipment_cost = excavator + hire

        # Subtotal
        subtotal = (
            material_cost +
            labour_cost +
            equipment_cost +
            electrical
        )

        # Charges
        transport = subtotal * self.transport_percent / 100
        safety = subtotal * self.safety_percent / 100
        gst = (subtotal + transport + safety) * self.gst_percent / 100
        profit = (subtotal + transport + safety + gst) * self.profit_percent / 100

        total = subtotal + transport + safety + gst + profit

        return {
            "Description": description,
            "Quantity": quantity,
            "Material Cost": round(material_cost, 2),
            "Labour Cost": round(labour_cost, 2),
            "Equipment Cost": round(equipment_cost, 2),
            "Transport": round(transport, 2),
            "Safety": round(safety, 2),
            "GST": round(gst, 2),
            "Profit": round(profit, 2),
            "Total Amount": round(total, 2)
        }


# Create Object
obj = QuotationCalculator()

# Calculate Quotation
result = obj.calculate(
    description="Brickwork",
    quantity=100,
    cement=542,
    sand=473,
    aggregate=0,
    steel=0,
    bricks=5000,
    water=300,
    excavator=0,
    hire=0,
    labour=3000,
    helper=2400,
    electrical=0
)

print("\n------ Construction Quotation ------")

for key, value in result.items():
    print(f"{key:<20}: {value}")
```

```python
import pandas as pd
#Used to save and load the trained machine learning model.
import joblib

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

#df becomes a table.
df = pd.read_csv("quotation.csv")
#Shows the first five rows.Useful to verify that the data loaded correctly.
print(df.head())
# Remove Missing Values in the dataset.
df.fillna(0, inplace=True)
#These are the input variables.
X = df[[
    "Item",
    "Quantity",
    "Cement_Bags",
    "Sand_CFT",
    "Aggregate_CFT",
    "Steel_Kg",
    "Bricks_Nos",
    "Water_Cost",
    "Equipment_Cost",
    "Labour_Cost",
    "Electrical_Cost",
    "Transport",
    "Safety",
    "GST",
    "Profit"
]]
#Target is an output,the value we want to predict.

y = df["Total"]
# Categorical-Item is text. & Numerical Columns

categorical_features = [
    "Item"
]
# Numerical Columns
numerical_features = [
    "Quantity",
    "Cement_Bags",
    "Sand_CFT",
    "Aggregate_CFT",
    "Steel_Kg",
    "Bricks_Nos",
    "Water_Cost",
    "Equipment_Cost",
    "Labour_Cost",
    "Electrical_Cost",
    "Transport",
    "Safety",
    "GST",
    "Profit"
]
   
#Different columns need different preprocessing.

#Example

#Item              → Text

#Quantity          → Number

#Steel_Kg          → Number

#ColumnTransformer applies

#OneHotEncoder for text
#Pass-through for numbers

preprocessor = ColumnTransformer(

    transformers=[

        (
            "cat",
            #Machine learning cannot understand text.To covert text to number, Onehotencoder helps.
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
#Means Quantity,Steel,Labour remain unchanged.
        (
            "num",
            "passthrough",
            numerical_features
        )

    ]

)

#Pipeline performs everything in one sequence.
#Instead of 
#Preprocess
#↓
#Train
#↓
#Predict
#Pipeline combines all into one model.

model = Pipeline([

    ("preprocessor", preprocessor),

    ("regressor",
#n_estimator-More trees(200) generally improve stability, but they also take longer to train.
     RandomForestRegressor(

        n_estimators=200,
#Keeps the results reproducible.If you rerun the code,the model gives similar results.
        random_state=42

    ))

])

# -----------------------------------
# Split Data
# -----------------------------------

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42

)

# -----------------------------------
# Train
# -----------------------------------

model.fit(

    X_train,

    y_train

)

# -----------------------------------
# Prediction
# -----------------------------------

prediction = model.predict(X_test)

# -----------------------------------
# Accuracy
# -----------------------------------

print()
#Average prediction error.
print("MAE :", mean_absolute_error(y_test, prediction))
#Measures error while giving more weight to large mistakes.Smaller values are better.
print("RMSE :", mean_squared_error(y_test, prediction) ** 0.5)

print("R2 Score :", r2_score(y_test, prediction))

#Used to save and load the trained machine learning model.
joblib.dump(

    model,

    "quotation_model.pkl"

)

print()

print("Model Saved Successfully")
```
