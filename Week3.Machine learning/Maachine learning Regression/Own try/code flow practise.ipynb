```python
Customer_names={"Sathya":{"Account_Number":159942616777, "balance":100000}, 
            "Kathir":{"Account_Number":150809201308,"balance":150000},
                      "Pramod":{"Account_Number":152503201325,"balance":75000}}
```

```python
print("Initital Account Details:")
for Customers, details in Customer_names.items():
    print(f"{Customers}-Account Number:{details['Account_Number']},Balance:{details['balance']}")
```

```python
for Customers, details in Customer_names.items():
    transaction_type=input(f"{Customers},Do you want ot Deposit(D) or withdraw(W)").upper()
    if transaction_type =="D":
        amount=float(input("Enter the Deposit Amount :$"))
        details["balance"]+=amount
        print(f"Deposit of ${amount} is successful, New balance:${details['balance']}")
    elif transaction_type =="W":
        amount=float(input("Enter the withdrawal Amount :$"))
        if amount <= details['balance']:
            details['balance']-= amount
            print(f"Withdrawal of ${amount} successful. New balance: ${details['balance']}")
        else:
            print("Insufficient funds. Withdrawal unsuccessful.")
    else:
        print("Invalid transaction type. Please enter 'D' for deposit or 'W' for withdrawal.")
    
```
