```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset=pd.read_csv("50_Startups.csv")
```

```python
dataset
```

```python
dataset=pd.get_dummies(dataset,dtype=int ,drop_first=True)
dataset
```

```python
dataset.columns
```

```python
Independent=dataset[['R&D Spend', 'Administration', 'Marketing Spend',
       'State_Florida', 'State_New York']]
```

```python
Independent
```

```python
dependent=dataset[["Profit"]]
```

```python
dependent
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(Independent,dependent,test_size=0.3,random_state=0)
```

```python
x_train
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV
param_grid={'n_estimators':[100,50],'criterion':['friedman_mse','squared_error','absolute_error','poisson']}
grid=GridSearchCV(RandomForestRegressor(),param_grid,refit=True,verbose=3,n_jobs=-1)
grid.fit(x_train,y_train)
```

```python
re=grid.cv_results_
print('R_score value for the best parameter{}:'.format(grid.best_params_))
```

```python
table=pd.DataFrame.from_dict(re)
```

```python
table
```

```python
RD_Spend_input = float(input("R&D_Spend: "))
Administration_input = float(input("Administration: "))
Marketing_Spend_input = float(input("Marketing_Spend: "))
State_Florida_input=int(input("State_Florida 0 or 1:"))
State_New_York_input=int(input("State_New York yes 0 or 1:"))
```

```python
Future_Prediction=grid.predict([[RD_Spend_input,Administration_input,Marketing_Spend_input,State_Florida_input,State_New_York_input]])
print("Future_Prediction {}:".format(Future_Prediction))
```
