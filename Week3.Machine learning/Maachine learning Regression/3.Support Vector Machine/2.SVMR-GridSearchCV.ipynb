```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset=pd.read_csv("50_Startups.csv")
```

```python
dataset
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
```

```python
dataset
```

```python
dataset.columns
```

```python
Independent=dataset[['RandD Spend', 'Administration', 'Marketing Spend',
       'State_Florida', 'State_New York']]
      
```

```python
Independent
```

```python
dependent=dataset[["Profit"]]
```

```python
dependent
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(Independent,dependent,test_size=0.30,random_state=0)
```

```python
x_train
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
x_train
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
param_grid={'C':[10,100,1000,2000,3000],'gamma':['auto','scale'],'kernel':['rbf','poly','sigmoid','linear']}
grid=GridSearchCV(SVR(),param_grid,refit=True,verbose=3,n_jobs=-1)
grid.fit(x_train,y_train)
```

```python
re=grid.cv_results_
print("R_score value for the best parameter {}:".format(grid.best_params_))
```

```python
table=pd.DataFrame.from_dict(re)
```

```python
Future_Prediction=grid.predict([[RandD_Spend_input,Administration_input,Marketing_Spend_input,State_Florida_input,State_New_York_input]])
print("Future_Prediction={}".format(Future_Prediction))
```
