```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset=pd.read_csv("Construction_Cost_Prediction_Sample.csv")
```

```python
dataset
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
```

```python
dataset
```

```python
dataset.columns
```

```python
Independent=dataset[['Project_ID', 'Area_sqft', 'Floors', 'Steel_kg', 'Cement_Bags',
       'Sand_cft', 'Labour_Days', 'Location_Factor', 'Project_Duration_Days',
        'Foundation_Type_Pile', 'Foundation_Type_Raft',
       'Structure_Type_PEB', 'Structure_Type_RCC']]
      
```

```python
Independent
```

```python
dependent=dataset[['Total_Cost_INR']]
```

```python
dependent
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
param_grid={'C':[100,200],'gamma':['auto','scale'],'kernel':['rbf']}
grid=GridSearchCV(SVR(),param_grid,refit=True,verbose=3,n_jobs=-1)
grid.fit(Independent,dependent)
```

```python
re=grid.cv_results_
print("R_score value for the best parameter {}:".format(grid.best_params_))
```

```python
table=pd.DataFrame.from_dict(re)
```

```python
Future_Prediction=grid.predict([[Project_ID_input,Area_sqft_input,Floors_input,Steel_kg_input,Cement_Bags_input,Sand_cft_input,Labour_Days_input,Location_Factor_input,Project_Duration_Days_input,Foundation_Type_Pile_input,Foundation_Type_Raft_input,Structure_Type_PEB_input,Structure_Type_RCC_input]])
print("Future_Prediction={}".format(Future_Prediction))
```
